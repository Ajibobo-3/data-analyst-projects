```python
def clean_currency(val):
    if pd.isna(val):
           return None


    val = str(val).replace('€', '').strip()

    if 'M' in val:
        return float(val.replace('M', '')) * 1000000

    elif 'K' in val:
        return float(val.replace('K', '')) * 1000

    try:
        return float(val)
    except ValueError:
        return None

df['Value_clean'] = df['Value'].apply(clean_currency)
df['Wage_clean'] = df['Wage'].apply(clean_currency)
```

```python
df
```

```python
df.drop(columns = ['Value_clean', 'Wage_clean'], inplace = True)
```

```python
def clean_height(val):
    if pd.isna(val):
       return None


    val = str(val).strip()
    if "'" in val:
        feet, inches = val.replace('"', '').split("'")
        feet = float(feet)
        inches = float(inches) if inches else 0.0
        return round((feet*30.48) + (inches*2.54))

    elif 'cm' in val:
        return float(val.replace('cm', '').strip())

    try:
        float(val)
    except ValueError:
        return None

df['Height_clean'] = df['Height'].apply(clean_height)
```

```python
df['Joined_clean'] = pd.to_datetime(df['Joined'], errors = 'coerce')
```

```python
df
```

```python
df['IR_clean'] = df['IR'].str.replace('★', '').str.strip().astype(int)
```

```python
def star_remover(val):
    if pd.isna(val):
        return None

    val = str(val).replace('★', '').strip()
    try:
        return int(val)
    except ValueError:
        return None

df['W/F_clean'] = df['W/F'].apply(star_remover)
df['SM_clean'] = df['SM'].apply(star_remover)
df['IR_clean'] = df['IR'].apply(star_remover)
```
